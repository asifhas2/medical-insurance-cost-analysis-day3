```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("insurance.csv")
df_clean = df.copy()

print(f"number of rows: {df_clean.shape[0]} and number of columns: {df_clean.shape[1]} ")
```

```python
df_clean.head()
df_clean.tail()
df_clean.sample(10)
```

```python
df_clean.shape
df_clean.columns
df_clean.info()
```

```python
df_clean.isnull().sum()
```

```python
# drop duplicat 

print(df_clean.duplicated().sum())
df_clean = df_clean.drop_duplicates()
print(df_clean.duplicated().sum())
```

```python
numerical_sumary = df_clean.describe()
```

```python
sns.pairplot(df_clean, hue='smoker', diag_kind = 'kid')
plt.show()
```

```python
print(f"Average insurance cost : ${round(numerical_sumary['charges'].mean(),2)}")
print(f"Average insurance cost : ${round(numerical_sumary['charges'].min(),2)}")
print(f"Average insurance cost : ${round(numerical_sumary['charges'].max(),2)}")
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df_clean['charges'], bins=30, kde= True)
plt.title('Distribution of Insurance charge ')
plt.xlabel('Charges ($)')
plt.ylabel('Number of people')
plt.show()
```

```python
plt.figure(figsize=(5,5))
sns.barplot(x='smoker', y = 'charges', data = df_clean , errorbar = None)
plt.title('Average Insurance charge by smoker status')
plt.xlabel(' Smoker')
plt.ylabel('Average Charge ($)')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x='region', y = 'charges', data = df_clean , errorbar = None)
plt.title('Average Insurance charge by region status')
plt.xlabel(' region')
plt.ylabel('Average Charge ($)')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x='age', y = 'charges', hue='smoker',data=df_clean)
plt.title('Average Insurance charge by age (colored by smoker status )')
plt.xlabel(' Age')
plt.ylabel('Average Charge ($)')
plt.show()
```
